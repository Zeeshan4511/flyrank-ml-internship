## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My lane is **Healthcare**, with a focus on **colorectal cancer**.

This is a **classification** problem because the goal is to predict whether a patient is likely to have colorectal cancer based on clinical and demographic information.

The model produces one of two possible outcomes:

- Positive (colorectal cancer)
- Negative (no colorectal cancer)

This prediction is intended to support healthcare professionals in identifying patients who may require further medical investigation.

```python
# ML task type
ml_task = "Binary Classification"

print("ML Task:", ml_task)
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target is **Cancer_Status**, which indicates whether colorectal cancer was observed for a patient.

Example values:

- 1 = Colorectal Cancer
- 0 = No Colorectal Cancer

The label comes from an **observed medical outcome**, such as a confirmed diagnosis after clinical evaluation or screening, rather than a manually defined rule.

If a confirmed diagnosis is not available, a validated screening result could be used as a proxy.

```python
import pandas as pd

# Update the path if your dataset is stored elsewhere
df = pd.read_csv("/content/crc_dataset.csv")

print(df.head())

print("\nColumns:")
print(df.columns.tolist())
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The primary success metric is **Recall (Sensitivity)**.

In colorectal cancer screening, failing to identify a patient who actually has cancer can delay diagnosis and treatment. Therefore, achieving a high recall is more important than maximising overall accuracy.

A strong model should also maintain reasonable Precision and F1 Score to reduce unnecessary follow-up investigations.

```python
success_metric = "Recall"

print("Primary Evaluation Metric:", success_metric)
```

The model learns from these patient-level observations to estimate the likelihood of colorectal cancer.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Colorectal cancer risk depends on many factors that interact in complex ways.

A simple rule such as:

"If Age > 50, then Cancer"

would ignore many important variables, including family history, lifestyle, laboratory results, symptoms, and other clinical measurements.

Machine Learning can learn patterns across many features at the same time and estimate risk based on combinations of observed information rather than relying on a single threshold.

The model is intended as a **decision-support tool** for healthcare professionals, not as a replacement for clinical judgement.
